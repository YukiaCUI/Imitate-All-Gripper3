```python
import h5py

def explore_hdf5(file_path):
    def print_attrs(name, obj):
        print(f"\nObject name: {name}")
        if isinstance(obj, h5py.Dataset):
            print(" - Type: Dataset")
            print(f" - Shape: {obj.shape}")
            print(f" - Dtype: {obj.dtype}")
        elif isinstance(obj, h5py.Group):
            print(" - Type: Group")
        # 打印属性
        for key, val in obj.attrs.items():
            print(f"   - Attr: {key} = {val}")

    with h5py.File(file_path, 'r') as f:
        print("HDF5 File Structure:")
        f.visititems(print_attrs)
        # action_data = f['action'][:10]
        # qpos_data = f['observations/qpos'][:10]
        # print("\nSample Action Data:")
        # print(action_data)
        # print("\nSample Qpos Data:")
        # print(qpos_data)


# 使用方法：将路径替换为你的 hdf5 文件路径
print("=================my file: ==================")
explore_hdf5("/media/cyj/DATA/大四下/Imitate-All/data/hdf5/gripper_task/episode_1.hdf5")

print("=================example file: ==================")
explore_hdf5("/media/cyj/DATA/大四下/Imitate-All/Data_Test/example_data.hdf5")
```

```python
import h5py
import numpy as np
import cv2

def main():
    filename = '/media/cyj/DATA/大四下/Imitate-All/Data_Test/episode_0.hdf5'
    with h5py.File(filename, 'r') as f:
        # 加载数据
        action = f['action'][:]
        qpos = f['observations/qpos'][:]
        images = f['observations/images/0'][:]

        # 输出 shape 信息
        print(f"Action shape: {action.shape}")   # (N, 8)
        print(f"Qpos shape: {qpos.shape}")       # (N, 8)
        print(f"Images shape: {images.shape}")   # (N, W*H*C)

        # 解码并展示图像及对应角度
        for i in range(len(qpos)):
            img_encoded = images[i]
            img = cv2.imdecode(img_encoded, cv2.IMREAD_COLOR)
            if img is not None:
                cv2.imshow('Frame', img)
                print(f"Frame {i}: Angles = {qpos[i]}")
                if cv2.waitKey(200) & 0xFF == ord('q'):
                    break
            else:
                print(f"Frame {i}: Failed to decode image.")

    cv2.destroyAllWindows()

if __name__ == '__main__':
    main()
```
